```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt('data_2.txt')
d = data[:, 0]
pd = data[:, 1]
bd = data[:, 2]
x=np.arange(1,101)
```

```python
plt.bar(x,d)
plt.xlabel("Value")
plt.ylabel("Probability")
plt.title("Histogram for Dice(100)")
```

```python
plt.bar(x,pd)
plt.xlabel("Value")
plt.ylabel("Probability")
plt.title("Histogram for PenaltyDice(100)")
```

```python
plt.bar(x,bd)
plt.xlabel("Value")
plt.ylabel("Probability")
plt.title("Histogram for BonusDice(100)")
```
